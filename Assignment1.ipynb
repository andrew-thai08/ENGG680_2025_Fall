## Part 6: Linear LSE
### (35 marks)

### Curve Fitting

curvefitting.png

### Observations: values of a function f(a) in range of -π to π, contaminated with random noise

### Unknown States: coefficients of polynomial functions for curve fitting

```python
# step 1
# import necessary libraries
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# step 2
# Load data.xlsx file, which contains the observations (with noise), variance of observations and true observations
# hint: use pandas library to parse excel file and use numpy library to convert it to array
# (1 mark)
#Load data file into pandas
df_data =  pd.read_excel('/content/drive/MyDrive/ENGG680_2025_Fall/Lab_Assignment1/data.xlsx')
#convert data to NumPy array
arr_data = df_data.to_numpy()

#store the noisy observations, variance of observations and true observations into numpy arrays seperately
# (1 mark)
#Observations with noise to numpy
l = df_data['l'].to_numpy()
#Variance of observations to numpy
l_var = df_data['l_var'].to_numpy()
#True observations to numpy
l_true = df_data['l_true'].to_numpy()
```

```python
# step 3
# define the number of observation n
# (1 mark)

n = len(l) #number of observations equals the length of observation array

# define the observation weight matrix P for LSE as identity matrix
# identity weight matrix simply means the observations will be unweighted in the following LSE
# (1 mark)
P = np.identity(n) #Observations will be unweighted

# define the range of function input(i.e., constants a), from start to end; and the numerical array of a
start = -1* np.pi
end = np.pi
a = np.linspace(start, end, n)
```

```python
# step 4
# use the matplotlib to plot the true and noisy observations together in one single figure
# use scatter for noisy observation plotting
# for the true observation, label it as "f(a)"" and set color as 'k'
# for the noisy observation scatter plot, label it as "f(a) + e", set color as 'tab:blue', set marker as 'o' and set alpha as 0.3
# include proper axis labels, legend and title
# scale the range of x-axis to be -1 to 1
# keep this format for the rest of the plots in this lab assignment
# (2 marks)

fig, ax = plt.subplots(nrows=1, ncols=1, figsize=(14, 5))

#Plot true observations
ax.plot(a,l_true,label="f(a)", color='k')
#Plot noisy observations
ax.scatter(a,l,label="f(a) + e", color='tab:blue', marker='o', alpha=0.3)
#Adding labels
ax.set_xlabel('a')
ax.set_ylabel('Observations')
ax.set_title('True vs. Noisy Observations')
ax.legend()
ax.set_xlim(-1,1)

ax.grid(True, which='both',ls='-',color='0.65')
fig.tight_layout()
plt.show()
```

```python
# step 5
# define a curve fitting function named fit_polynomial_curve
# the function should implement the linear least square estimation
# the function should include all important intermediate steps in LSE
# the input and output of the funciton are described in the comment below in red.
# please keep code clean and well-commented
# (7 marks)

def fit_polynomial_curve(a: np.ndarray, l: np.ndarray, P: np.ndarray, deg: int) -> tuple:
    '''
    Args:
        a:              function input (constants)
        l:              noisy observations
        P:              observation weight matrix
        deg:            degree of polynomial to fit to data
    Returns:
        x:              estimated polynomial coefficients
        cov_x:          uncertainty in estimated coefficients
        apv:            a posterier variance factor (goodness of fit)
        v:              observation residuals
    '''

# Matrix A
    A = np.vstack([a**i for i in range(deg + 1)]).T

    # Weighted least squares estimation
    A_T_P_A = A.T @ P @ A
    A_T_P_l = A.T @ P @ l

    # Calculating the coefficient
    x = np.linalg.solve(A_T_P_A, A_T_P_l)

    # Residual calculation
    v = l - A @ x

    # Calcualting the covariance
    cov_x = np.linalg.inv(A_T_P_A) * np.var(v)

    # Calculating posterier variance factor
    apv = np.var(v)


    return (x, cov_x, apv, v)
```

```python
# step 6
# call the fit_polynomial_curve function and fetch the results
# use polynomial degrees of: 2, 4, 6
# (1 marks)

results = {}
for degree in [2, 4, 6]:
    x, cov_x, apv, residuals = fit_polynomial_curve(a, l, P, degree)

    #storing the results for future steps
    results[degree] = {
        'coefficients': x,
        'covariance': cov_x,
        'posterior_variance': apv,
        'residuals': residuals
    }

    print(f"Degree: {degree}")
    print("Estimated polynomial coefficient:", x)
    print("Uncertainty in estimated coefficients:", cov_x)
    print("Posterior variance factor (goodness of fit):", apv)
    print("Observation Residuals:", residuals)
    print()
```

```python
# step 7
# use the estimated coefficients x and constants array a, construct polynomial functions for each degree
# hint: use polyval function from numpy library (check input argument carefully)
# (1 marks)

quadratic_uw = np.polyval(results[2]['coefficients'], a)
quartic_uw = np.polyval(results[4]['coefficients'], a)
hexic_uw = np.polyval(results[6]['coefficients'], a)
```

```python
# step 8
# use the matplotlib to plot each polynomial function, noisy observations l, true observations l_true, together in one single figure
# for the polynomials of line, cubic and quintic, use color of orange, green and red respectively.
# (2 marks)

fig, ax = plt.subplots(nrows=1, ncols=1, figsize=(14, 5))

# noisy observation, blue from step 4 above
ax.scatter(a, l, label='Noisy Observations (l)', color='tab:blue', marker='o', alpha=0.5)

# true observation, k color from step 4 above
ax.plot(a, l_true, label='True Observations (l_true)', color='k', linewidth=2)

# polynomial plotting
ax.plot(a, quadratic_uw, label='Quadratic Fit (Degree 2)', color='orange')
ax.plot(a, quartic_uw, label='Quartic Fit (Degree 4)', color='green')
ax.plot(a, hexic_uw, label='Hexic Fit (Degree 6)', color='red')

# configuring the chart
ax.set_xlabel('input a')
ax.set_ylabel('Observations')
ax.set_xlim(-4, 4)
ax.set_ylim(-200,200)
ax.legend()
ax.set_title('Polynomial Functions, Noisy Observations and True Observations')

ax.grid(True, which='both',ls='-',color='0.65')

fig.tight_layout()

plt.show()
```

```python
# step 9
# print the a posterier variance factor for each polynomial function
# (3 marks)
for degree in [2, 4, 6]:
  apv = results[degree]['posterior_variance']

  print(f'The posterior variance factor for degree {degree}: {apv}')
```

```python
# step 10
# compute and print the norm between the predicted observations (output from each polynomial function) and the true observations
# (3 marks)
for degree in [2, 4, 6]:
  coefficients = results[degree]['coefficients']
  predicted = np.polyval(coefficients[::-1], a)

  norm = np.linalg.norm(predicted - l_true)

  print(f'The norm between the predicted observation and true observation for each degree {degree}: {norm}')
```

```python
# step 12
# now, let's define the P matrix with the proper weights of observation
# use the variance of observation l_var, define P matrix
# (1 mark)
import numpy as np
P = np.diag(1/l_var)
```
